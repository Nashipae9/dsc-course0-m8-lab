## Exploratory Data Analysis  
- Load in the cleaned data

```python
import pandas as pd

# Load the cleaned aviation dataset
df = pd.read_csv("data/AviationData_Cleaned.csv")

# Check the dataset size
print("Dataset shape:", df.shape)

# Display the first five rows
df.head()
```

```python
# Convert accident dates into datetime
df["Event.Date"] = pd.to_datetime(
    df["Event.Date"],
    errors="coerce"
)

# Check the earliest and latest dates
print("Earliest date:", df["Event.Date"].min())
print("Latest date:", df["Event.Date"].max())
```

```python
# FILTER ACCIDENTS FROM 1983 TO 2023

df = df[
    (df["Event.Date"].dt.year >= 1983) &
    (df["Event.Date"].dt.year <= 2023)
].copy()

# Check the results
print("Remaining records:", len(df))

print("Earliest date:", df["Event.Date"].min())
print("Latest date:", df["Event.Date"].max())
```

```python
#INSPECT AIRCRAFT CATEGORIES

print("Aircraft categories:")
print(df["Aircraft.Category"].value_counts(dropna=False))

print("\nProfessional build status:")
print(df["Amateur.Built"].value_counts(dropna=False))
```
